```python
import numpy as np 
import matplotlib.pyplot as plt 

x = np.linspace(-5,5,30)
x1 = np.linspace(0.01,5,30)

f1 = lambda x : x**2
f2 = lambda x : np.exp(x)
f3 = lambda x1 : np.log(x1)
f4 = lambda x : 1/(1+np.exp(-x))

# plot the four functions 
fig,ax = plt.subplots(2,2,figsize=(13,10))

ax[0,0].plot(x,f1(x))
ax[0,0].set_title("Function x^2")

ax[0,1].plot(x,f2(x))
ax[0,1].set_title("Function e^x")

ax[1,0].plot(x1,f3(x1))
ax[1,0].set_title("Function log(x)")

ax[1,1].plot(x,f4(x))
ax[1,1].set_title("Function 1/(1+e^(-x)) ")

plt.tight_layout()
plt.show()

```

```python
def numerical_derivative(f,x,h=1e-5):
    return (f(x+h)-f(x-h))/(2*h)

f5 = lambda x : np.sin(x)

# f1(x) = x^2
print(numerical_derivative(f1,0))
# f2(x) = e^x 
print(numerical_derivative(f2,0))
# f3(x) = log(x)
print(numerical_derivative(f3,1))
# f4(x) = 1/(1+e^(-x))
print(numerical_derivative(f4,0))
# f5(x) = sin(x)
print(numerical_derivative(f5,0))
```

```python
g = lambda x : x**2 + 3*x + 2
dg = lambda x: 2*x +3

x0 = 2
y0 = g(x0) # 12
slope = dg(x0) # 7

def tangent(x):
    return y0 + slope* (x-x0)

x = np.linspace(-5,5,200)
plt.figure(figsize=(12,8))
plt.plot(x,g(x),label="f(x) = x^2 + 3x + 2")
plt.plot(x,tangent(x),label="tangent at x =2")
plt.scatter(x0,y0,color="red",zorder=3)
plt.axhline(0, color="black", linewidth=0.5)
plt.axvline(0, color="black", linewidth=0.5)
plt.legend()
plt.grid(True)
plt.show()





```
